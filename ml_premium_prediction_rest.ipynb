### Feature Engineering

```python
df2.medical_history.unique()
```

```python
risk_scores = {
    'diabetes':6,
    'heart disease':8,
    'high blood pressure':6,
    'thyroid':5,
    'no disease':0,
    'none':0
}


df2[['disease1', 'disease2']] = df2['medical_history'].str.split(" & ", expand=True).apply(lambda x: x.str.lower())
df2.head(3)
```

```python
df2['disease1'] = df2['disease1'].fillna('none')
df2['disease2'] =  df2['disease2'].fillna('none')
df2['total_risk_score'] = 0

for disease in ['disease1','disease2']:
    df2['total_risk_score'] += df2[disease].map(risk_scores)

min_score = df2['total_risk_score'].min()
max_score = df2['total_risk_score'].max()

df2['normalized_risk_score']= (df2.total_risk_score - min_score) / (max_score - min_score)

df2.head()
```

```python
df2.insurance_plan.unique()
```

```python
df2['insurance_plan'] = df2['insurance_plan'].map({
    'Bronze':1,
    'Silver':2,
    'Gold':3
})

df2['insurance_plan'].unique()
```

```python
df2.income_level.unique()
```

```python
df2['income_level'] = df2['income_level'].map({
    '<10L':1,
    '10L - 25L':2,
    '> 40L':3,
    '25L - 40L':4
})

df2['income_level'].unique()
```

```python
df2.head()
```

```python
nominal_cols = ['gender', 'region', 'marital_status', 'bmi_category', 'smoking_status', 'employment_status']
df3 = pd.get_dummies(df2, columns=nominal_cols, drop_first=True, dtype=int)
df3.head(5)
```

```python
df4 = df3.drop(['medical_history', 'disease1','disease2','total_risk_score'], axis=1)
df4.head()
```

```python
corr_matrix = df4.corr()

plt.figure(figsize=(20,12))
sns.heatmap(corr_matrix, annot=True)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df4.head(2)
```

```python
X = df4.drop('annual_premium_amount',axis='columns')
y = df4['annual_premium_amount']

from sklearn.preprocessing import MinMaxScaler
cols_to_scale = ['age', 'number_of_dependants', 'income_level', 'income_lakhs', 'insurance_plan']
scaler = MinMaxScaler()

X[cols_to_scale] = scaler.fit_transform(X[cols_to_scale])
X.describe()
```

```python
def calculate_vif(data):
    vif_df = pd.DataFrame()
    vif_df['Column'] = data.columns
    vif_df['VIF'] = [variance_inflation_factor(data.values, i) for i in range(data.shape[1])]
    return vif_df
```

```python
calculate_vif(X)
```

### Model Training

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=10)

print('X_train :',X_train.shape)
print('X_test :',X_test.shape)
print('y_train :',y_train.shape)
print('y_test :',y_test.shape)
```

```python
model_lr = LinearRegression()
model_lr.fit(X_train, y_train)
test_score = model_lr.score(X_test, y_test)
train_score = model_lr.score(X_train, y_train)
train_score, test_score
```

```python
y_pred = model_lr.predict(X_test)

mse_lr = mean_squared_error(y_test, y_pred)
rmse_lr = np.sqrt(mse_lr)
print(f"Linear Regression -->  MSE : {mse_lr}, RMSE : {rmse_lr}")
```

```python
X_reduced = X.drop('income_level', axis='columns')
X_reduced.head(2)
```

```python
model_lr.coef_
```

```python
model_lr.intercept_
```

```python
feature_importances = model_lr.coef_

coef_df = pd.DataFrame(feature_importances, index=X_test.columns, columns=['Coefficients'])
coef_df = coef_df.sort_values(by='Coefficients', ascending=True)

coef_df

plt.figure(figsize=(8,4))
plt.barh(coef_df.index, coef_df['Coefficients'], color='steelblue')
plt.xlabel('Coefficient Value')
plt.title('Feature Importance in Linear Regression')
plt.show()
```

```python
model_ridge = Ridge(alpha=0)
model_ridge.fit(X_train, y_train)
train_score = model_ridge.score(X_train, y_train)
test_score = model_ridge.score(X_test, y_test)
train_score, test_score
```

```python
from xgboost import XGBRegressor

model_xgb = XGBRegressor()
model_xgb.fit(X_train, y_train)
train_score = model_xgb.score(X_train, y_train)
test_score = model_xgb.score(X_test, y_test)
train_score, test_score
```

```python
y_pred = model_xgb.predict(X_test)

mse_lr = mean_squared_error(y_test, y_pred)
rmse_lr = np.sqrt(mse_lr)
print(f"XGBoost Regression -->  MSE : {mse_lr}, RMSE : {rmse_lr}")
```

```python
param_grid = {
    'n_estimators' : [20, 40 ,50],
    'learning_rate' : [0.01, 0.1, 0.2],
    'max_depth' : [3, 4, 5]
}

random_search = RandomizedSearchCV(model_xgb, param_grid, n_iter=10, cv=3, scoring='r2', random_state=42)
random_search.fit(X_train, y_train)
random_search.best_score_
```

```python
random_search.best_params_
```

```python
best_model = random_search.best_estimator_
```

```python
feature_importances = best_model.feature_importances_

coef_df = pd.DataFrame(feature_importances, index=X_test.columns, columns=['Coefficients'])
coef_df = coef_df.sort_values(by='Coefficients', ascending=True)

plt.figure(figsize=(8,4))
plt.barh(coef_df.index, coef_df['Coefficients'], color='steelblue')
plt.xlabel('Coefficient Value')
plt.title('Feature Importance in XGBoost')
plt.show()
```

### Error Analysis 

```python
y_pred = best_model.predict(X_test)

residuals = y_pred - y_test
residuals_pct = residuals*100/y_test

results_df = pd.DataFrame({
    'actual' : y_test,
    'predicted' : y_pred,
    'diff' : residuals,
    'diff_pct' : residuals_pct
})

results_df.head()
```

```python
sns.histplot(results_df['diff_pct'], kde=True)
```

```python
extreme_error_threshold = 10
extreme_results_df = results_df[np.abs(results_df.diff_pct) > extreme_error_threshold]
extreme_results_df.head()
```

```python
extreme_errors_df = extreme_results_df.shape[0] * 100 / results_df.shape[0]
extreme_errors_df
```

```python
results_df[np.abs(results_df.diff_pct) > 50].sort_values('diff_pct', ascending=False)
```
